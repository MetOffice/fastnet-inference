```python
ordered_vars: list[str] = []
for var, level in forecast_vars_order_fastnet:
    short_name = ERA5_VARIABLE_TO_SHORTHAND_NAME[var]
    if level != 0:
        short_name += f"_{level}"
    ordered_vars.append(short_name)

for var, level in nonforecast_vars_order_fastnet:
    short_name = ERA5_VARIABLE_TO_SHORTHAND_NAME[var]
    if level != 0:
        short_name += f"_{level}"
    ordered_vars.append(short_name)

print("In Anemoi, not in FastNet vars:")
print(set(ds.variables) - set(ordered_vars))
print("In FastNet vars, not in Anemoi:")
print(set(ordered_vars) - set(ds.variables))
```
